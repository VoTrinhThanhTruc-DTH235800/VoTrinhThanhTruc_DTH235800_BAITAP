Python hỗ trợ nhiều kiểu dữ liệu cơ bản, bao gồm:

*   **Kiểu số (Numeric Types):**
    *   `int` (Số nguyên): Ví dụ: 1, 100, -5
    *   `float` (Số thực): Ví dụ: 1.0, 100.5, -5.2
    *   `complex` (Số phức): Ví dụ: 1 + 2j
*   **Kiểu chuỗi (Sequence Types):**
    *   `str` (Chuỗi): Ví dụ: "hello", 'world'

    *   `set` (Tập hợp): Tập hợp không có thứ tự, không chứa các phần tử trùng lặp. Ví dụ: {1, 2, 3}

*   **Kiểu ánh xạ (Mapping Types):**
    *   `dict` (Từ điển): Tập hợp các cặp khóa-giá trị không có thứ tự. Ví dụ: {'a': 1, 'b': 2}
*   **Kiểu Boolean (Boolean Type):**
    *   `bool`: Đại diện cho giá trị đúng (`True`) hoặc sai (`False`).
*   **Kiểu None (None Type):**
    *   `NoneType`: Đại diện cho giá trị rỗng hoặc không có gì.

Câu 6: Trình bày ý nghĩa toán tử /, //, %, **, and, or, is.


---


*   `/` (Toán tử chia): Thực hiện phép chia và trả về kết quả là số thực (float).
    *   Ví dụ: `10 / 3` kết quả là `3.333...`
*   `//` (Toán tử chia lấy phần nguyên): Thực hiện phép chia và trả về phần nguyên của kết quả.
    *   Ví dụ: `10 // 3` kết quả là `3`
*   `%` (Toán tử chia lấy phần dư - Modulo): Thực hiện phép chia và trả về số dư của phép chia.
    *   Ví dụ: `10 % 3` kết quả là `1`
*   `**` (Toán tử lũy thừa): Thực hiện phép tính lũy thừa (số mũ).
    *   Ví dụ: `2 ** 3` kết quả là `8` (2 mũ 3)
*   `and` (Toán tử logic AND): Trả về `True` nếu cả hai biểu thức đều đúng, ngược lại trả về `False`.
    *   Ví dụ: `True and False` kết quả là `False`
*   `or` (Toán tử logic OR): Trả về `True` nếu ít nhất một trong hai biểu thức đúng, ngược lại trả về `False`.
    *   Ví dụ: `True or False` kết quả là `True`
*   `is` (Toán tử nhận dạng): Kiểm tra xem hai biến có tham chiếu đến cùng một đối tượng trong bộ nhớ hay không. Trả về `True` nếu chúng cùng tham chiếu, ngược lại trả về `False`.
    *   Ví dụ:

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]
print(a is b) # Kết quả: True (a và b tham chiếu cùng một đối tượng)
print(a is c) # Kết quả: False (a và c là hai đối tượng khác nhau, dù có nội dung giống nhau)
```

Cách phổ biến nhất để nhập dữ liệu từ bàn phím trong Python là sử dụng hàm `input()`.

*   **`input()`**: Hàm này đọc một dòng từ bàn phím (chuẩn đầu vào), chuyển dòng đó thành một chuỗi (string), và trả về chuỗi đó.
    *   Bạn có thể truyền một chuỗi gợi ý (prompt) vào `input()` để hiển thị cho người dùng biết cần nhập gì.
    *   Ví dụ:

```python
# Nhập một chuỗi từ bàn phím
ten = input("Nhập tên của bạn: ")
print("Xin chào, " + ten)
```

*   **Chuyển đổi kiểu dữ liệu**: Hàm `input()` luôn trả về một chuỗi. Nếu bạn muốn nhập các kiểu dữ liệu khác như số nguyên (int) hoặc số thực (float), bạn cần chuyển đổi chuỗi trả về sang kiểu dữ liệu mong muốn bằng cách sử dụng các hàm như `int()`, `float()`.
    *   Ví dụ:

```python
# Nhập một số nguyên từ bàn phím và chuyển đổi sang kiểu int
tuoi_str = input("Nhập tuổi của bạn: ")
try:
    tuoi = int(tuoi_str)
    print("Tuổi của bạn là:", tuoi)
except ValueError:
    print("Đầu vào không hợp lệ. Vui lòng nhập một số nguyên.")

# Nhập một số thực từ bàn phím và chuyển đổi sang kiểu float
chieu_cao_str = input("Nhập chiều cao của bạn (mét): ")
try:
    chieu_cao = float(chieu_cao_str)
    print("Chiều cao của bạn là:", chieu_cao, "mét")
except ValueError:
    print("Đầu vào không hợp lệ. Vui lòng nhập một số.")
```

Câu 8: Trình bày các loại lỗi khi lập trình và cách bắt lỗi trong Python.


---


Trong lập trình Python, có hai loại lỗi chính:

1.  **Lỗi cú pháp (Syntax Errors):** Xảy ra khi code không tuân thủ đúng các quy tắc ngữ pháp của Python. Trình thông dịch sẽ phát hiện lỗi này trước khi chương trình thực thi.
    *   Ví dụ: Thiếu dấu hai chấm (`:`) sau câu lệnh `if`, `for`, `while`, `try`, `except`; sai tên hàm hoặc từ khóa; thiếu dấu ngoặc đơn/ngoặc kép.

2.  **Lỗi ngoại lệ (Exceptions):** Xảy ra khi cú pháp của code đúng nhưng có vấn đề trong quá trình thực thi. Python sẽ dừng chương trình và báo lỗi ngoại lệ.
    *   Ví dụ: Chia cho 0 (`ZeroDivisionError`), truy cập chỉ mục không tồn tại trong danh sách (`IndexError`), sử dụng khóa không tồn tại trong từ điển (`KeyError`), cố gắng chuyển đổi một giá trị không hợp lệ sang một kiểu dữ liệu khác (`ValueError`), lỗi khi nhập/xuất file (`IOError`), lỗi về tên biến/hàm chưa được định nghĩa (`NameError`).

**Cách bắt lỗi (Exception Handling) trong Python:**

Python sử dụng khối `try`, `except`, và `finally` để xử lý lỗi ngoại lệ.

*   **`try`**: Khối mã mà bạn nghi ngờ có thể xảy ra lỗi.
*   **`except`**: Khối mã được thực thi khi một lỗi ngoại lệ xảy ra trong khối `try`. Bạn có thể chỉ định loại lỗi cụ thể muốn bắt.
*   **`finally`**: Khối mã này (nếu có) sẽ luôn được thực thi, bất kể có lỗi xảy ra hay không. Nó thường được dùng để dọn dẹp tài nguyên (ví dụ: đóng file).

Ví dụ:

```python
try:
    # Khối mã có thể gây ra lỗi
    so_nguyen = int(input("Nhập một số nguyên: "))
    ket_qua = 10 / so_nguyen
    print("Kết quả:", ket_qua)

except ValueError:
    # Xử lý lỗi khi người dùng nhập không phải số nguyên
    print("Lỗi: Vui lòng nhập một số nguyên hợp lệ.")

except ZeroDivisionError:
    # Xử lý lỗi khi người dùng nhập số 0
    print("Lỗi: Không thể chia cho 0.")

except Exception as e:
    # Xử lý các loại lỗi khác (nếu có)
    print(f"Đã xảy ra lỗi không xác định: {e}")

finally:
    # Khối này luôn được thực thi
    print("Kết thúc chương trình xử lý lỗi.")
```

Câu 9: Giải thích kết quả tính toán của các biểu thức


---



```python
# Cho các biến
i1 = 2
i2 = 5
i3 = -3
d1 = 2.0
d2 = 5.0
d3 = -0.5

# thực hiện phép tính từ trái qua phải, nhân chia trước cộng trừ sau
# và ưu tiên thực hiện các phép tín trong ngoặc
# / (Toán tử thực hiện phép chia): thực hiện phép chia trả về kết quả là số thực
# // (Toán tử chia lấy phần nguyên): Thực hiện phép chia và trả về phần nguyên của kết quả.



# (a) i1 + (i2 * i3)
result_a = i1 + (i2 * i3)
print(f"(a) {i1} + ({i2} * {i3}) = {result_a}")

# (b) i1 * (i2 + i3)
result_b = i1 * (i2 + i3)
print(f"(b) {i1} * ({i2} + {i3}) = {result_b}")

# (c) i1 / (i2 + i3)
result_c = i1 / (i2 + i3)
print(f"(c) {i1} / ({i2} + {i3}) = {result_c}")

# (d) i1 // (i2 + i3)
result_d = i1 // (i2 + i3)
print(f"(d) {i1} // ({i2} + {i3}) = {result_d}")

# (e) i1 / i2 + i3
result_e = i1 / i2 + i3
print(f"(e) {i1} / {i2} + {i3} = {result_e}")

# (f) i1 // i2 + i3
result_f = i1 // i2 + i3
print(f"(f) {i1} // {i2} + {i3} = {result_f}")

# (g) 3 + 4 + 5 / 3
result_g = 3 + 4 + 5 / 3
print(f"(g) 3 + 4 + 5 / 3 = {result_g}")

# (h) 3 + 4 + 5 // 3
result_h = 3 + 4 + 5 // 3
print(f"(h) 3 + 4 + 5 // 3 = {result_h}")

# (i) (3 + 4 + 5) / 3
result_i = (3 + 4 + 5) / 3
print(f"(i) (3 + 4 + 5) / 3 = {result_i}")

# (j) (3 + 4 + 5) // 3
result_j = (3 + 4 + 5) // 3
print(f"(j) (3 + 4 + 5) // 3 = {result_j}")

# (k) d1 + (d2 * d3)
result_k = d1 + (d2 * d3)
print(f"(k) {d1} + ({d2} * {d3}) = {result_k}")

# (l) d1 + d2 * d3
result_l = d1 + d2 * d3
print(f"(l) {d1} + {d2} * {d3} = {result_l}")

# (m) d1 / d2 - d3
result_m = d1 / d2 - d3
print(f"(m) {d1} / {d2} - {d3} = {result_m}")

# (n) d1 / (d2 - d3)
result_n = d1 / (d2 - d3)
print(f"(n) {d1} / ({d2} - {d3}) = {result_n}")

# (o) d1 + d2 + d3 / 3
result_o = d1 + d2 + d3 / 3
print(f"(o) {d1} + {d2} + {d3} / 3 = {result_o}")

# (p) (d1 + d2 + d3) / 3
result_p = (d1 + d2 + d3) / 3
print(f"(p) ({d1} + {d2} + {d3}) / 3 = {result_p}")

# (q) d1 + d2 + (d3 / 3)
result_q = d1 + d2 + (d3 / 3)
print(f"(q) {d1} + {d2} + ({d3} / 3) = {result_q}")

# (r) 3 * (d1 + d2) * (d1 - d3)
result_r = 3 * (d1 + d2) * (d1 - d3)
print(f"(r) 3 * ({d1} + {d2}) * ({d1} - {d3}) = {result_r}")
```
